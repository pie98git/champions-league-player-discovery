```python
import pandas as pd

df = pd.read_csv("../data/raw/players_data_light-2025_2026.csv")

df = df[df["Min"] > 500].copy()

ucl_teams = [
    "Arsenal", "Aston Villa", "Barcelona", "Bayern Munich", "Dortmund",
    "Chelsea", "Eintracht Frankfurt", "Inter", "Juventus", "Leverkusen",
    "Liverpool", "Manchester City", "Marseille", "Milan", "Monaco",
    "Napoli", "Newcastle United", "Paris Saint-Germain", "RB Leipzig",
    "Real Madrid", "Tottenham Hotspur", "Union Berlin", "Villarreal"
]

df["is_ucl_team"] = df["Squad"].isin(ucl_teams)

striker_features = ["Gls", "Ast", "G+A", "Sh", "SoT", "G/Sh"]

strikers = df[df["Pos"] == "FW"].copy()

striker_data = strikers[
    ["Player", "Squad", "Comp", "is_ucl_team"] + striker_features
]

striker_data.head()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors
```

```python
X = striker_data[striker_features]
```

```python
model = NearestNeighbors(
    n_neighbors=6,
    metric="euclidean"
)

model.fit(X_scaled)
```

```python
def find_similar_non_ucl_players(player_name, n_neighbors=10):
    player_match = striker_data[striker_data["Player"] == player_name]
    
    if player_match.empty:
        return f"Player '{player_name}' not found."
    
    player_index = player_match.index[0]
    
    distances, indices = model.kneighbors(
        [X_scaled[player_index]],
        n_neighbors=n_neighbors
    )
    
    similar_players = striker_data.iloc[indices[0]].copy()
    similar_players["distance"] = distances[0]
    
    hidden_targets = similar_players[
        similar_players["is_ucl_team"] == False
    ].copy()
    
    return hidden_targets
```

```python
# Rebuild striker dataset with reliability thresholds and per90 features

df = df[
    (df["MP"] >= 15) &
    (df["Min"] >= 900)
].copy()

strikers = df[df["Pos"] == "FW"].copy()

striker_features = ["Gls", "Ast", "G+A", "Sh", "SoT", "G/Sh"]

striker_data = strikers[
    [
        "Player",
        "Squad",
        "Comp",
        "MP",
        "Min",
        "90s",
        "is_ucl_team"
    ] + striker_features
].copy()

striker_data["Gls_per90"] = striker_data["Gls"] / striker_data["90s"]
striker_data["Ast_per90"] = striker_data["Ast"] / striker_data["90s"]
striker_data["Sh_per90"] = striker_data["Sh"] / striker_data["90s"]
striker_data["SoT_per90"] = striker_data["SoT"] / striker_data["90s"]

advanced_striker_features = [
    "Gls_per90",
    "Ast_per90",
    "Sh_per90",
    "SoT_per90",
    "G/Sh"
]

striker_data = striker_data.reset_index(drop=True)

X = striker_data[advanced_striker_features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = NearestNeighbors(
    n_neighbors=10,
    metric="euclidean"
)

model.fit(X_scaled)

striker_data.head()
```
